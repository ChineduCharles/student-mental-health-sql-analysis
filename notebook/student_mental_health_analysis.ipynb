## Extended Analysis: Data Understanding and Quality Assessment

Before extending the analysis, I examined the dataset to understand its
structure, identify incomplete or unusual records, and determine which
observations should be included in the analysis.

This step is important because conclusions about student mental health
should be based only on valid student responses.

### Are all rows actual student records?

```python
SELECT
    COUNT(*) AS total_rows,

    COUNT(*) FILTER (
        WHERE inter_dom IN ('Inter', 'Dom')
    ) AS valid_student_records,

    COUNT(*) FILTER (
        WHERE TRIM(COALESCE(inter_dom, '')) = ''
    ) AS non_student_rows

FROM students;
```

### Data Quality Finding: Non-Student Rows

The raw dataset contains 286 rows, but only 268 represent individual
student responses.

The remaining 18 rows have blank values for `inter_dom` and contain
summary or aggregate values rather than complete respondent-level
records. These rows are therefore not valid student observations.

**Decision:** The original dataset will be preserved unchanged, while
all subsequent student-level analysis will be restricted to records
where `inter_dom` is either `Inter` or `Dom`.

### Missing Value Assessment

After identifying the valid student records, I checked whether important
demographic, mental-health, and help-seeking variables contained missing
values that could affect the analysis.

```python
WITH valid_students AS (
    SELECT *
    FROM students
    WHERE inter_dom IN ('Inter', 'Dom')
),

column_values AS (
    SELECT
        j.key AS column_name,
        j.value AS column_value
    FROM valid_students s
    CROSS JOIN LATERAL jsonb_each_text(to_jsonb(s)) AS j(key, value)
)

SELECT
    column_name,
    COUNT(*) AS total_records,

    COUNT(*) FILTER (
        WHERE column_value IS NULL
           OR TRIM(column_value) = ''
    ) AS missing_values,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE column_value IS NULL
               OR TRIM(column_value) = ''
        )
        / COUNT(*),
        2
    ) AS missing_percentage

FROM column_values
GROUP BY column_name
ORDER BY missing_values DESC, column_name;
```

### Missing-Value Findings

A comprehensive assessment of all variables identified missing values in
only two columns among the 268 valid student records:

- `internet`: 26 missing values (9.70%)
- `intimate`: 8 missing values (2.99%)

All other variables, including the main mental-health measures
(`todep`, `tosc`, and `toas`), are complete.

No records were removed at this stage. The missing values will be
investigated further and handled only when analyses involving these
specific variables are performed.

```python
SELECT
    inter_dom,
    academic,
    gender,
    age,
    intimate
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
  AND (
      intimate IS NULL
      OR TRIM(intimate) = ''
  );
```

```python
SELECT
    inter_dom,
    dep,
    internet,
    internet_bi
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
  AND (
      internet IS NULL
      OR TRIM(CAST(internet AS TEXT)) = ''
  );
```

### Missing-Intimate Values

The 8 missing values in `intimate` occur only among international
students. Because there is no reliable information from which to infer
their relationship status, these values will not be imputed. The
students will remain in the dataset and will only be excluded from any
analysis that specifically requires `intimate`.

The 26 missing values in the numerical `internet` variable require
further investigation because the corresponding binary variable,
`internet_bi`, is still populated. I therefore examined the relationship
between the original numerical score and its binary representation
before deciding how the missing values should be handled.

```python
SELECT
    internet,
    internet_bi,
    COUNT(*) AS student_count
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY internet, internet_bi
ORDER BY internet NULLS LAST, internet_bi;
```

### Missing-Internet Values

The numerical `internet` variable contains 26 missing responses.

For students with a recorded numerical score, the corresponding binary
variable (`internet_bi`) follows a consistent rule:

- Scores 1–4 are classified as `No`
- Scores 5–7 are classified as `Yes`

However, all 26 records with a missing numerical `internet` value are
also classified as `No` in `internet_bi`.

Because the original numerical response is unavailable, these 26 cases
cannot confidently be interpreted as genuine `No` responses.

**Decision:** Analyses using the numerical `internet` variable will use
only students with a recorded score. If a binary Internet help-seeking
measure is required, it will be derived from the observed numerical
scores rather than treating missing responses as `No`.

### Duplicate Handling Decision

The dataset does not contain a unique respondent identifier, so duplicate
responses cannot be verified reliably.

For this analysis, the 268 valid student records are retained as provided
in the source dataset. No records are removed on the basis of assumed
duplication.

## 2. Student Population Profile

Before examining mental-health outcomes, I first explored the composition
of the survey sample. Understanding who is represented in the data
provides important context for interpreting later comparisons.

```python
SELECT
    inter_dom,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) / SUM(COUNT(*)) OVER (),
        2
    ) AS percentage
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom
ORDER BY student_count DESC;
```

### Finding

International students make up 75% of the survey sample, while domestic
students account for 25%. Because the two groups are unequal in size,
later comparisons should consider both the average outcome and the
number of students represented in each group.

```python
SELECT
    inter_dom,
    gender,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY inter_dom),
        2
    ) AS percentage_within_group
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom, gender
ORDER BY inter_dom, student_count DESC;
```

### Finding

The gender distribution is similar across both student groups.

- Domestic students: 62.69% female and 37.31% male
- International students: 63.68% female and 36.32% male

Because the gender proportions are nearly identical, differences observed
later between international and domestic students are less likely to be
explained simply by differences in gender composition.

### What is the academic-level composition of each student group?

```python
SELECT
    inter_dom,
    academic,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY inter_dom),
        2
    ) AS percentage_within_group
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom, academic
ORDER BY inter_dom, student_count DESC;
```

### Finding

Both student groups are predominantly undergraduate. However, graduate
students make up 9.95% of the international sample compared with only
1.49% of the domestic sample.

This difference in academic composition should be kept in mind when
comparing mental-health outcomes between international and domestic
students, since the two groups are not identical in their academic-level
distribution.

### How does age differ between international and domestic students?

```python
SELECT
    inter_dom,
    COUNT(*) AS student_count,
    ROUND(AVG(age), 0) AS average_age,
    MIN(age) AS minimum_age,
    MAX(age) AS maximum_age
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom
ORDER BY inter_dom;
```

```python
SELECT
    inter_dom,
    age_cate,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY inter_dom),
        2
    ) AS percentage_within_group
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom, age_cate
ORDER BY inter_dom, age_cate;
```

### Finding

The average ages of international and domestic students are relatively
similar at 21.03 and 20.40 years respectively.

However, international students have a broader age range (17–31 years)
than domestic students (18–27 years) and are more represented in the
higher age categories.

Together with the academic-level results, this suggests that the
international sample includes a somewhat larger share of older and
graduate students. These compositional differences will be considered
when interpreting later comparisons between the two groups.

### Population Profile Summary

The survey sample consists primarily of international, female, and
undergraduate students. International and domestic students have very
similar gender distributions and broadly similar average ages.

The main compositional differences are the larger sample size of
international students and their greater representation among graduate
and older students. These differences provide context for the
mental-health comparisons that follow.

## 3. Mental Health Profile

After understanding the composition of the survey population, the next
step is to examine the mental-health measures captured in the dataset.

The analysis focuses primarily on three measures:

- **PHQ (`todep`)** – depression score
- **SCS (`tosc`)** – social connectedness score
- **ASISS (`toas`)** – acculturative stress score

Higher PHQ scores indicate greater depressive symptoms, higher SCS
scores indicate stronger social connectedness, and higher ASISS scores
indicate greater acculturative stress.

```python
SELECT
    COUNT(*) AS student_count,
    ROUND(AVG(todep), 2) AS average_phq,
    ROUND(AVG(tosc), 2) AS average_scs,
    ROUND(AVG(toas), 2) AS average_acculturative_stress
FROM students
WHERE inter_dom IN ('Inter', 'Dom');
```

```python
SELECT
    depsev AS depression_severity,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentage
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY depsev
ORDER BY
    CASE depsev
        WHEN 'Min' THEN 1
        WHEN 'Mild' THEN 2
        WHEN 'Mod' THEN 3
        WHEN 'ModSev' THEN 4
        WHEN 'Sev' THEN 5
    END;
```

## 4. International vs Domestic Students

After examining the overall mental-health profile of the survey sample,
I compared international and domestic students to determine whether the
two groups show differences in depression, social connectedness, and
acculturative stress.

### How do the average mental-health scores differ between international and domestic students?

```python
SELECT
    inter_dom,
    COUNT(*) AS student_count,
    ROUND(AVG(todep), 2) AS average_phq,
    ROUND(AVG(tosc), 2) AS average_scs,
    ROUND(AVG(toas), 2) AS average_acculturative_stress
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom
ORDER BY inter_dom;
```

```python
SELECT
    inter_dom,
    depsev AS depression_severity,
    COUNT(*) AS student_count,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY inter_dom),
        2
    ) AS percentage_within_group
FROM students
WHERE inter_dom IN ('Inter', 'Dom')
GROUP BY inter_dom, depsev
ORDER BY
    inter_dom,
    CASE depsev
        WHEN 'Min' THEN 1
        WHEN 'Mild' THEN 2
        WHEN 'Mod' THEN 3
        WHEN 'ModSev' THEN 4
        WHEN 'Sev' THEN 5
    END;
```

### Finding

The depression-severity distribution does not show international students
consistently experiencing more severe depressive symptoms than domestic
students.

Domestic students have a somewhat larger proportion of respondents in
the moderate, moderately severe, and severe categories, while
international students have a larger proportion in the minimal and mild
categories.

This suggests that differences between international and domestic
students may be more complex than depression severity alone and should
be examined alongside social connectedness and acculturative stress.

## 5. Length of Stay and Mental Health Among International Students

The original DataCamp analysis examined whether the length of time
international students had lived in Japan was associated with their
mental-health scores.

In this extended analysis, I revisit that question while also considering
the number of students represented at each length of stay. This is
important because averages based on very small groups may not be
representative.

```python
SELECT
    stay,
    COUNT(*) AS student_count,
    ROUND(AVG(todep), 2) AS average_phq,
    ROUND(AVG(tosc), 2) AS average_scs,
    ROUND(AVG(toas), 2) AS average_acculturative_stress
FROM students
WHERE inter_dom = 'Inter'
GROUP BY stay
ORDER BY stay;
```

### Finding

Among the better-represented stay groups (1–4 years), acculturative stress
generally increases with length of stay, while social connectedness tends
to decline. Depression scores also rise between years 1 and 3, although
the pattern is not strictly linear.

However, the number of students drops sharply after four years. Several
later stay groups contain only one respondent, making their averages too
unstable for reliable comparison.

For this reason, exact-year averages are interpreted cautiously and a
broader stay-category analysis is used to obtain a more stable view of
the relationship.


```python
SELECT
    stay_cate,
    COUNT(*) AS student_count,
    ROUND(AVG(todep), 2) AS average_phq,
    ROUND(AVG(tosc), 2) AS average_scs,
    ROUND(AVG(toas), 2) AS average_acculturative_stress
FROM students
WHERE inter_dom = 'Inter'
GROUP BY stay_cate
ORDER BY
    CASE stay_cate
        WHEN 'Short' THEN 1
        WHEN 'Medium' THEN 2
        WHEN 'Long' THEN 3
    END;
```

### Finding

When length of stay is grouped into broader categories, a clearer pattern
emerges.

Average acculturative stress increases from 72.80 among short-stay
students to 78.81 among long-stay students, while average social
connectedness declines from 38.11 to 35.57.

Depression scores do not follow the same linear pattern. PHQ increases
from 7.48 in the short-stay group to 8.72 in the medium-stay group before
declining to 7.86 among long-stay students.

Overall, the results suggest that longer residence is more consistently
associated with higher acculturative stress and lower social
connectedness than with steadily increasing depressive symptoms.

The extended analysis produced four main findings:

1. **The survey sample is dominated by international students.**  
   Of the 268 valid student records, 75% are international students and

2. **Depression severity does not show a simple international-versus-domestic pattern.**  
   Domestic students have a somewhat larger proportion of respondents in

3. **Exact-year comparisons of length of stay can be misleading.**  
   International students who had stayed for more than four years were

4. **Broader stay categories reveal a clearer pattern.**  
   Among international students, average acculturative stress increased
   from 72.80 for short-stay students to 78.81 for long-stay students,
   while social connectedness declined from 38.11 to 35.57.

   Depression scores did not increase consistently with length of stay,
   suggesting that longer residence is more clearly associated with

   increasing depressive symptoms.

## 7. Limitations and Conclusion

### Limitations

This analysis is based on survey data from one university and therefore
should not be generalized to all international or domestic students.

The dataset is cross-sectional, so the observed relationships cannot be
interpreted as causal effects.

Some length-of-stay groups contained very few students, which made
exact-year averages unstable. Broader stay categories were therefore used
to support more reliable comparisons.

The dataset also contained 18 non-student summary rows, which were
excluded from respondent-level analysis. Missing values were found in
the `internet` and `intimate` variables and were handled only when those
variables were relevant to an analysis.

### Conclusion

The analysis suggests that the mental-health experience of international
students is more complex than a simple relationship between time spent
abroad and depression.

While depressive symptoms did not increase consistently with length of
stay, longer-stay international students showed higher average
acculturative stress and lower social connectedness.

This highlights an important analytical distinction: remaining in a host
country for longer does not necessarily imply easier adjustment.

More broadly, the project demonstrates why sample size, data quality and
the structure of grouped data should be considered before interpreting
averages or drawing conclusions from survey data.
